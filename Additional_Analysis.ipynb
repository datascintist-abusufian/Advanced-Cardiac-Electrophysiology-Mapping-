#APD80 Mapping

Extraction APD80 (Action Potential Duration at 80% repolarization) and compare automated vs. manual detection.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

def compute_apd80(image_series, threshold=0.8):
    """
    Compute APD80 based on intensity changes over time.
    Args:
        image_series (numpy array): Shape (X, Y, T) where T is time.
        threshold (float): 80% repolarization threshold.
    Returns:
        APD80 map
    """
    X, Y, T = image_series.shape
    apd_map = np.zeros((X, Y))

    for i in range(X):
        for j in range(Y):
            signal = image_series[i, j, :]
            max_val = np.max(signal)
            min_val = np.min(signal)
            apd_threshold = min_val + threshold * (max_val - min_val)

            try:
                apd_index = np.where(signal <= apd_threshold)[0][0]
                apd_map[i, j] = apd_index
            except IndexError:
                apd_map[i, j] = np.nan  # No clear repolarization

    return apd_map

# Compute APD80 using automated detection
apd80_automated = compute_apd80(images, threshold=0.8)

# Simulate a manual APD80 detection method (e.g., using different fixed thresholds)
apd80_manual = compute_apd80(images, threshold=0.7)  # Simulated manual threshold

# Plot side-by-side comparison
fig, axes = plt.subplots(1, 2, figsize=(6, 4))

# Automated APD80 Map
sns.heatmap(apd80_automated, cmap="viridis", cbar=True, ax=axes[0])
axes[0].set_title("APD80 Mapping (Automated)")

# Manual APD80 Map
sns.heatmap(apd80_manual, cmap="viridis", cbar=True, ax=axes[1])
axes[1].set_title("APD80 Mapping (Manual)")

plt.tight_layout()
plt.show()
```

#Conduction Velocity Vector Analysis
This estimates conduction velocity by computing spatial derivatives of the signal.

```python
from scipy.ndimage import sobel

def compute_conduction_velocity(image_series):
    """
    Compute conduction velocity using gradient-based estimation.
    Args:
        image_series (numpy array): Shape (X, Y, T).
    Returns:
        Velocity field (X, Y).
    """
    X, Y, T = image_series.shape
    cv_map = np.zeros((X, Y))

    # Compute spatial gradients
    grad_x = sobel(image_series[:, :, T//2], axis=0)  # Midpoint frame
    grad_y = sobel(image_series[:, :, T//2], axis=1)

    # Compute conduction velocity
    cv_map = np.sqrt(grad_x**2 + grad_y**2)

    return cv_map

# Compute conduction velocity map
cv_map = compute_conduction_velocity(images)

# Plot conduction velocity vectors
plt.figure(figsize=(6, 5))
sns.heatmap(cv_map, cmap="coolwarm", cbar=True)
plt.title("Conduction Velocity Map")
plt.show()
```

#Localization of Conduction Slowing (normEA)
normalize conduction velocity to highlight regions of slowing.

```python
def compute_normea(cv_map):
    """
    Normalize conduction velocity to highlight slowing regions.
    Args:
        cv_map (numpy array): Conduction velocity map.
    Returns:
        Normalized conduction velocity.
    """
    normea = (cv_map - np.min(cv_map)) / (np.max(cv_map) - np.min(cv_map))
    return normea

# Compute normEA
normea_map = compute_normea(cv_map)

# Plot normalized conduction slowing
plt.figure(figsize=(6, 5))
sns.heatmap(normea_map, cmap="magma", cbar=True)
plt.title("Localization of Conduction Slowing (normEA)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.ndimage import gaussian_gradient_magnitude

def compute_apd80(image_series, threshold=0.8):
    """
    Compute APD80 based on intensity changes over time.
    Args:
        image_series (numpy array): Shape (X, Y, T) where T is time.
        threshold (float): 80% repolarization threshold.
    Returns:
        APD80 map (X, Y)
    """
    X, Y, T = image_series.shape
    apd_map = np.zeros((X, Y))

    for i in range(X):
        for j in range(Y):
            signal = image_series[i, j, :]
            max_val = np.max(signal)
            min_val = np.min(signal)
            apd_threshold = min_val + threshold * (max_val - min_val)

            try:
                apd_index = np.where(signal <= apd_threshold)[0][0]
                apd_map[i, j] = apd_index
            except IndexError:
                apd_map[i, j] = np.nan  # No clear repolarization

    return apd_map

def compute_conduction_velocity(image_series):
    """
    Compute conduction velocity using spatial gradients in intensity.
    Args:
        image_series (numpy array): Shape (X, Y, T) where T is time.
    Returns:
        Conduction velocity map (X, Y)
    """
    X, Y, T = image_series.shape
    velocity_map = np.zeros((X, Y))

    for t in range(T - 1):  # Temporal gradient
        frame_diff = np.abs(image_series[:, :, t + 1] - image_series[:, :, t])
        spatial_gradient = gaussian_gradient_magnitude(frame_diff, sigma=1)
        velocity_map += spatial_gradient

    return velocity_map / (T - 1)  # Average across time

# Load the entire dataset (assuming images is the full dataset)
# Split into baseline and arrhythmia conditions
baseline_images = images[:, :, :2500]  # First 2500 frames
arrhythmia_images = images[:, :, 2500:]  # Last 2500 frames

# Compute APD80 maps
apd80_baseline = compute_apd80(baseline_images)
apd80_arrhythmia = compute_apd80(arrhythmia_images)

# Compute conduction velocity
cv_baseline = compute_conduction_velocity(baseline_images)
cv_arrhythmia = compute_conduction_velocity(arrhythmia_images)

# Generate Visualizations
fig, axes = plt.subplots(2, 2, figsize=(8, 6))

# APD80 Heatmaps
sns.heatmap(apd80_baseline, cmap="viridis", ax=axes[0, 0])
axes[0, 0].set_title("APD80 Mapping - Baseline")

sns.heatmap(apd80_arrhythmia, cmap="inferno", ax=axes[0, 1])
axes[0, 1].set_title("APD80 Mapping - Arrhythmia")

# Conduction Velocity Distributions
sns.histplot(cv_baseline.flatten(), bins=30, color='blue', kde=True, ax=axes[1, 0])
axes[1, 0].set_title("Conduction Velocity Distribution - Baseline")

sns.histplot(cv_arrhythmia.flatten(), bins=30, color='red', kde=True, ax=axes[1, 1])
axes[1, 1].set_title("Conduction Velocity Distribution - Arrhythmia")

plt.tight_layout()
plt.show()
```

```python
import scipy.io
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Define paths to example automated and manual conduction velocity data (placeholders)
automated_cv_file = "/content/06062018_A_clean/mouseventricle06062018 A_FLEC Baseline Paced.mat"
manual_cv_file = "/content/06062018_A_clean/mouseventricle06062018 A_FLEC Baseline SR.mat"

# Function to load CV data from .mat files
def load_cv_data(mat_file):
    if os.path.exists(mat_file):
        mat_data = scipy.io.loadmat(mat_file)
        # Assuming conduction velocity is stored under a key like 'cv_data'
        for key in mat_data.keys():
            if "cv" in key.lower() or "velocity" in key.lower():
                return np.array(mat_data[key]).flatten()
    return None

# Load automated and manual CV data
automated_cv = load_cv_data(automated_cv_file)
manual_cv = load_cv_data(manual_cv_file)

# Ensure valid data before plotting
if automated_cv is not None and manual_cv is not None:
    plt.figure(figsize=(8, 6))

    # Plot histogram with KDE
    sns.histplot(automated_cv, bins=30, kde=True, color="blue", label="Automated CV", alpha=0.6)
    sns.histplot(manual_cv, bins=30, kde=True, color="red", label="Manual CV", alpha=0.6)

    plt.xlabel("Conduction Velocity (cm/s)")
    plt.ylabel("Frequency")
    plt.title("Comparison of Automated vs. Manual Conduction Velocity")
    plt.legend()
    plt.show()
else:
    print("Conduction velocity data not found in the specified .mat files.")
```
